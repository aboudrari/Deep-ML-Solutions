***Matrix-Vector Dot Product***

```python
# using numpy 

import numpy as np
def matrix_dot_vector(a: list[list[int|float]], b: list[int|float]) -> list[int|float]:
	# Return a list where each element is the dot product of a row of 'a' with 'b'.
	# If the number of columns in 'a' does not match the length of 'b', return -1. 
	a = np.array(a)
	b = np.array(b)
	if a.shape[1] != b.shape[0]:          # a.shape[0] → number of rows   ||   a.shape[1] → number of columns
		return -1
	return (a @ b).tolist()	

matrix_dot_vector(a = [[1, 2], [2, 4]], b = [1, 2])
```

***Transpose of a Matrix***

```python
def transpose_matrix(a: list[list[int|float]]) -> list[list[int|float]]:
	"""
	Transpose a 2D matrix by swapping rows and columns.
	
	Args:
		a: A 2D matrix of shape (m, n)
	
	Returns:
		The transposed matrix of shape (n, m)
	"""
	# Your code here
	m , n = len(a) , len(a[0])
	transposed_a = [[0] * m for _  in range(n)]
	for i in range(n):
		for j in range(m):
			transposed_a[i][j] = a[j][i]
	return transposed_a        

	# one line of code 
	# return [ list(row) for row in zip(*a) ]      

transpose_matrix(a = [[1, 2, 3], [4, 5, 6]])
```

***Calculate Mean by Row or Column***

```python
def calculate_matrix_mean(matrix: list[list[float]], mode: str) -> list[float]:
	n_columns = len(matrix[0])
	means = []
	if mode == 'row':
		for row in matrix:
			mean = sum(row) / len(row)
			means.append(mean)
	else:
		for i in range(n_columns):
			column = [row[i] for row in matrix]
			mean = sum(column) / len(column)
			means.append(mean)		
	return means

calculate_matrix_mean(matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]], mode = 'column')
```

***Calculate Cosine Similarity Between Vectors***

Cosine Similarity explains how similar two vectors are in terms of direction.
The result is always between -1 and 1:
	1.0  → vectors point in the same direction (perfect similarity)
  0.0  → vectors are perpendicular (no similarity)
 -1.0  → vectors point in opposite directions

How it works step by step:
 Step 1 - Validate inputs:
	- Convert both inputs to numpy arrays
	- Check if either vector is empty
	- Check if both vectors have the same shape

Step 2 - Dot Product (numerator):
	- Multiply each pair of corresponding elements together
	- Example: v1=[1,2,3] v2=[2,4,6] → (1×2) + (2×4) + (3×6) = 28

Step 3 - Magnitude (denominator):
	- Calculate the length of each vector using np.linalg.norm()
	- Which is the square root of the sum of squared elements
	- Then multiply both magnitudes together

Step 4 - Divide:
	- dot_product / (magnitude_v1 * magnitude_v2)
	- This gives the cosine of the angle between the two vectors

```python
import numpy as np

def cosine_similarity(v1, v2):
	v1 = np.array(v1)
	v2 = np.array(v2)
	
	if len(v1) == 0 or len(v2) == 0:
		raise ValueError("Vectors cannot be empty")
	if v1.shape != v2.shape:
		raise ValueError("Vectors must have the same shape")
	
	dot_product = np.dot(v1, v2)
	magnitude = np.linalg.norm(v1) * np.linalg.norm(v2)
	
	return dot_product / magnitude
	
	
v1 = np.array([1, 2, 3])
v2 = np.array([2, 4, 6])
print(round(cosine_similarity(v1, v2), 3))	
	
```

***Calculate 2x2 Matrix Inverse***

```python
def inverse_2x2(matrix: list[list[float]]) -> list[list[float]] | None:
	"""
	Calculate the inverse of a 2x2 matrix.
	
	Args:
		matrix: A 2x2 matrix represented as [[a, b], [c, d]]
	
	Returns:
		The inverse matrix as a 2x2 list, or None if the matrix is singular
		(i.e., determinant equals zero)
	"""
	# Your code here
	# so first extract a,b,c,d : remember that matrix is list of list 
	a , b , c , d  = matrix[0][0] , matrix[0][1] , matrix[1][0] , matrix[1][1]
	det = a * d - b * c
	if det == 0:
		return None
	
	# So each element gets multiplied by 1/det, and a and d swap, while b and c get negated.

	return [[d/det, -b/det], [-c/det, a/det]]
	 # return [[(1/det) * d,  (1/det) * -b] , [(1/det) * -c, (1/det) * a]]


print(inverse_2x2([[4, 7], [2, 6]]))
```

***Calculate Eigenvalues of a Matrix***

```python
import math
def calculate_eigenvalues(matrix: list[list[float|int]]) -> list[float]:
	
	a , b , c , d  = matrix[0][0] , matrix[0][1] , matrix[1][0] , matrix[1][1]
	det = a * d - b * c
	trace = a + d
	lambda1 = (trace + math.sqrt(trace**2 - 4*det)) / 2
	lambda2 = (trace - math.sqrt(trace**2 - 4*det)) / 2
 
	return sorted([lambda1, lambda2], reverse=True)

print(calculate_eigenvalues([[2, 1], [1, 2]]))
	
```

***Matrix times Matrix***

```python
def matrixmul(a:list[list[int|float]],b:list[list[int|float]])-> list[list[int|float]]:
	# Two matrices align for multiplication when: 
		# The number of columns in matrix A equals the number of rows in matrix B
	if len(a[0]) != len(b):
		return -1
	rows = len(a)
	cols = len(b[0])
	c = [[0] * cols for _ in range(rows)]
	 
	for i in range(len(a)):        # rows of A
		for j in range(len(b[0])): # columns of B
			for k in range(len(b)): # accumulate
				c[i][j] += a[i][k] * b[k][j]

	return c

print(matrixmul([[1,2,3],[2,3,4],[5,6,7]],[[3,2,1],[4,3,2],[5,4,3]]))
```

***Calculate Covariance Matrix***

```python
def calculate_covariance_matrix(vectors: list[list[float]]) -> list[list[float]]:
	# Your code here
	f1 , f2 = vectors[0] , vectors[1]
	mean1 = sum(f1) / len(f1)
	mean2 = sum(f2) / len(f2)
	cove_f1_f2 = sum(((f1[i]-mean1) * (f2[i]-mean2))  for i in range(len(f1))) / (len(f1)-1)
	cov_f1_f1 = sum(((f1[i]-mean1)**2)  for i in range(len(f1))) / (len(f1)-1)
	cov_f2_f2  = sum(((f2[i]-mean2)**2)  for i in range(len(f1))) / (len(f1)-1)
	# cov(f2,f1) == cov(f1,f2)
	cov_f2_f1 = cove_f1_f2
	
	return [[cov_f1_f1,cove_f1_f2],[cov_f2_f1,cov_f2_f2]]
	

print(calculate_covariance_matrix([[1, 2, 3], [4, 5, 6]]))    
	
```

```python
def calculate_covariance_matrix(vectors: list[list[float]]) -> list[list[float]]:
	n = len(vectors)        # number of features
	m = len(vectors[0])     # number of observations
	
	# step 1: calculate mean for each feature
	means = [sum(v) / len(v) for v in vectors]
	
	# step 2: initialize n x n matrix of zeros
	cov_matrix = [[0]*n for _ in range(n)]
	
	# step 3: nested loop over each pair (i, j)
	for i in range(n):
		for j in range(n):
			cov_matrix[i][j] = sum((vectors[i][k] - means[i]) * (vectors[j][k] - means[j]) for k in range(m)) / (m - 1)
	
	return cov_matrix

print(calculate_covariance_matrix([[1, 2, 3], [4, 5, 6]]))
```

***Poisson Distribution Probability Calculator***

```python
import math

def poisson_probability(k, lam):
	"""
	Calculate the probability of observing exactly k events in a fixed interval,
	given the mean rate of events lam, using the Poisson distribution formula.
	:param k: Number of events (non-negative integer)
	:param lam: The average rate (mean) of occurrences in a fixed interval
	"""
	# Your code here
	# val = (math.exp(-lam) * lam ** k) / math.factorial(k)
	val = (math.pow(lam,k) * math.exp(-lam) ) / math.factorial(k)
	return round(val,5)

print(poisson_probability(3, 5))
```

***Binomial Distribution Probability***

```python
import math

def binomial_probability(n: int, k: int, p: float) -> float:
	"""
	Calculate the probability of exactly k successes in n Bernoulli trials.
	
	Args:
		n: Total number of trials
		k: Number of successes
		p: Probability of success on each trial
	
	Returns:
		Probability of k successes
	"""
	# Your code here
	binomal_coeff = math.factorial(n) / (math.factorial(k) * math.factorial(n-k))
	Success_term = math.pow(p,k)
	Failure_term = math.pow((1-p),(n-k))
	Result =  binomal_coeff * Success_term * Failure_term 
	return Result

print(round(binomial_probability(6, 2, 0.5), 5))

```

***Descriptive Statistics Calculator***

```python
import numpy as np
from collections import Counter

def descriptive_statistics(data: list | np.ndarray) -> dict:
	"""
	Calculate various descriptive statistics metrics for a given dataset.
	
	Args:
		data: List or numpy array of numerical values
	
	Returns:
		Dictionary containing mean, median, mode, variance, standard deviation,
		percentiles (25th, 50th, 75th), and interquartile range (IQR)
	"""
	# Your code here
	count = Counter(data).most_common(1)
	mean  = np.mean(data)
	median = np.median(data)
	mode = count[0][0]
	variance = np.var(data)
	std_dev = np.std(data)
	per_25 = np.percentile(data,25)
	per_50 = np.percentile(data,50)
	per_75 = np.percentile(data,75)
	interquartile_range = per_75 - per_25

	return {
	'mean': round(float(mean), 4),
	'median': round(float(median), 4),
	'mode': mode,
	'variance': round(float(variance), 4),
	'standard_deviation': round(float(std_dev), 4),
	'25th_percentile': round(float(per_25), 4),
	'50th_percentile': round(float(per_50), 4),
	'75th_percentile': round(float(per_75), 4),
	'interquartile_range': round(float(interquartile_range), 4)
}


result = descriptive_statistics([1, 2, 2, 3, 4, 4, 4, 5])
print({k: round(v, 4) if isinstance(v, float) else v for k, v in result.items()})
```

***Calculate the Phi Coefficient***

```python
def phi_corr(x: list[int], y: list[int]) -> float:
	x00 = 0
	x01 = 0
	x10 = 0
	x11 = 0
	for xi, yi in zip(x, y):
		if xi == 0 and yi == 0:
			x00 += 1
		elif xi == 0 and yi == 1:
			x01 += 1
		elif xi == 1 and yi == 0:
			x10 += 1
		else:
			x11 += 1
	val = ((x00 * x11) - (x01 * x10)) / ((x00+x01)*(x10+x11)*(x00+x10)*(x01+x11)) ** 0.5
	return round(val, 4)

print(phi_corr([1, 1, 0, 0], [0, 0, 1, 1]))
```

***Gradient Direction and Magnitude***


```python
import numpy as np

def gradient_direction_magnitude(gradient: list) -> dict:
	"""
	Calculate the magnitude and direction of a gradient vector.
	
	Args:
		gradient: A list representing the gradient vector
	
	Returns:
		Dictionary containing:
		- magnitude: The L2 norm of the gradient
		- direction: Unit vector in direction of steepest ascent
		- descent_direction: Unit vector in direction of steepest descent
	"""
	# Your code here
	gradient = np.array(gradient)
	Magnitude = np.linalg.norm(gradient)
	if Magnitude == 0 :
		Magnitude = None
	Direction = gradient / Magnitude
	Descent_direction = -Direction
	return {'magnitude': Magnitude, 'direction': Direction.tolist(), 'descent_direction': Descent_direction.tolist()}

result = gradient_direction_magnitude([3.0, 4.0])
print(f"{result['magnitude']:.4f},{[round(d,4) for d in result['direction']]},{[round(d,4) for d in result['descent_direction']]}")
```

***Product Rule for Derivatives***

```python
import numpy as np
def product_rule_derivative(f_coeffs, g_coeffs):
	def poly_derivative(coeffs):
		return [i * coeffs[i] for i in range(1, len(coeffs))]
	
	f_prime = poly_derivative(f_coeffs)
	g_prime = poly_derivative(g_coeffs)
	
	f_prime_g = np.convolve(f_prime, g_coeffs)
	f_g_prime = np.convolve(f_coeffs, g_prime)
	
	# Add the two results (hint: np.polyadd handles different lengths!)
	result = np.polyadd(f_prime_g,f_g_prime)
	
	# Round to 4 decimal places and convert to list
	result = [round(float(c), 4) for c in result]
	
	# Remove trailing zeros (but return [0.0] if all zeros)
	while len(result) > 1 and result[-1] == 0.0:
		result.pop()
	
	return result

print(product_rule_derivative([1, 2], [3, 4]))
```

***Implement Gradient Descent Variants with MSE Loss***

```python
import numpy as np

def gradient_descent(X, y, weights, learning_rate, n_epochs, batch_size=1, method='batch'):
	"""
	Perform gradient descent optimization.
	
	Args:
		X: Feature matrix of shape (m, n)
		y: Target values of shape (m,)
		weights: Initial weights of shape (n,)
		learning_rate: Step size for gradient descent
		n_epochs: Number of complete passes through the dataset
		batch_size: Size of batches for mini-batch gradient descent (default: 1)
		method: Type of gradient descent ('batch', 'stochastic', or 'mini_batch')
	
	Returns:
		Optimized weights
	"""
	# Your code here
	X = np.array(X)
	y = np.array(y)
	for _ in range(n_epochs):
		if method == 'batch':
			prediction = X @ weights
			error = prediction - y
			gradient = 2 * X.T @ error / len(X)
			weights = weights - learning_rate * gradient
		elif method == 'stochastic':
			for i in range(len(X)):
				X_sample = X[i:i+1]  # shape (1, n)
				y_sample = y[i:i+1]  # shape (1,)
				prediction = X_sample @ weights
				error = prediction - y_sample
				gradient = 2 * X_sample.T @ error 
				weights = weights - learning_rate * gradient
		elif method == 'mini_batch':
			for i in range(0, len(X), batch_size):
				X_batch = X[i : i + batch_size]
				y_batch = y[i : i + batch_size]   
				prediction = X_batch @ weights
				error = prediction - y_batch
				gradient = 2 * X_batch.T @ error / batch_size
				weights = weights - learning_rate * gradient

	return weights        



X = np.array([[1, 1], [2, 1], [3, 1], [4, 1]])
y = np.array([2, 3, 4, 5])
weights = np.zeros(X.shape[1])
learning_rate = 0.01
n_epochs = 100

# Test Batch Gradient Descent
output = gradient_descent(X, y, weights, learning_rate, n_epochs, method='batch')
print(output)
```

***Chain Rule for Composite Functions***

```python
import numpy as np

def compute_chain_rule_gradient(functions: list[str], x: float) -> float:
	"""
	Compute derivative of composite functions using chain rule.
	
	Args:
		functions: List of function names (applied right to left)
				  Available: 'square', 'sin', 'exp', 'log'
		x: Point at which to evaluate derivative
	
	Returns:
		Derivative value at x
	
	Example:
		['sin', 'square'] represents sin(x²)
		['exp', 'sin', 'square'] represents exp(sin(x²))
	"""
	# Your code here
	# Forward pass 
	derivs = {
	'square': lambda x: 2*x,
	'sin': np.cos,
	'exp': np.exp,
	'log': lambda x: 1/x,
}
	funcs = {
	'square': lambda x: x**2,
	'sin': np.sin,
	'exp': np.exp,
	'log': np.log,
}
	u = x
	intermediates = [x]
	for f in reversed(functions):
		u = funcs[f](u)
		intermediates.append(u)
	
	# back pass 
	result = 1.0
	for i , f in enumerate(reversed(functions)):
		result *= derivs[f](intermediates[i])
	return round(result , 6)
		
result = compute_chain_rule_gradient(['sin', 'square'], 1.0); print(f"{result:.6f}")        
		
	   

	
```

***Implement F-Score Calculation for Binary Classification***

```python
import numpy as np

def f_score(y_true, y_pred, beta):
	"""
	Calculate F-Score for a binary classification task.

	:param y_true: Numpy array of true labels
	:param y_pred: Numpy array of predicted labels
	:param beta: The weight of precision in the harmonic mean
	:return: F-Score rounded to three decimal places
	"""
	TP = np.sum((y_true == 1 ) & (y_pred == 1))
	TN = np.sum((y_true == 0 ) & (y_pred == 0))
	FN = np.sum((y_true == 1 ) & (y_pred == 0))
	FP = np.sum((y_true == 0 ) & (y_pred == 1))
	
	Recall = TP / (TP + FN )
	Precision = TP / (TP + FP)
	if (beta**2 * Precision) + Recall == 0:
		return 0.0
	F_Score = ( 1 + beta**2) * ((Precision * Recall) / ((beta**2 * Precision) + Recall) )
	return np.round(F_Score , 3)



y_true = np.array([1, 0, 1, 1, 0, 1])
y_pred = np.array([1, 0, 1, 0, 0, 1])
beta = 1
print(f_score(y_true, y_pred, beta))
```

```python
def law_of_total_probability(priors: dict, conditionals: dict) -> float:
	"""
	Compute P(A) using the Law of Total Probability.
	
	Args:
		priors: Dictionary mapping partition event names to P(Bi)
		conditionals: Dictionary mapping partition event names to P(A|Bi)
	
	Returns:
		float: The total probability P(A), rounded to 4 decimal places
	"""
	# Your code here
	result = 0
	for key  in priors.keys() :
		probofB = priors[key]
		probAofB = conditionals[key]
		result += probofB * probAofB
	return result

priors = {'B1': 0.3, 'B2': 0.7}
conditionals = {'B1': 0.2, 'B2': 0.5}
print(law_of_total_probability(priors, conditionals))    
```

***Implement K-Fold Cross-Validation***

```python
import numpy as np
from typing import List, Tuple

def k_fold_cross_validation(n_samples: int, k: int = 5, shuffle: bool = True) -> List[Tuple[List[int], List[int]]]:
	"""
	Generate train/test index splits for k-fold cross-validation.
	
	Args:
		n_samples: Total number of samples in the dataset
		k: Number of folds (default 5)
		shuffle: Whether to shuffle indices before splitting (default True)
	
	Returns:
		List of (train_indices, test_indices) tuples
	"""
	# Your code here
	indices = np.arange(n_samples) 
	if shuffle :
		np.random.shuffle(indices)
	folds = []
	start = 0
	base_size = n_samples // k
	remainder = n_samples %  k 
	for i in range(k):
		if i < remainder:
			fold_size = base_size + 1
		else:
			fold_size = base_size 
		end = start + fold_size
		folds.append(indices[start:end])
		start = end
	result = []
	for i in range(k):
		test_indices  = folds[i]        # just fold i
		train_indices = np.concatenate([folds[j] for j in range(k) if j != i])         # all other folds combined
		result.append((train_indices, test_indices))
	return result
		
import numpy as np
np.random.seed(42)
print(k_fold_cross_validation(n_samples=10, k=5, shuffle=False))
```

```python
import numpy as np 

def matrix_determinant_and_trace(matrix: list[list[float]]) -> tuple[float, float]:
		
		def determinant(matrix):
			if len(matrix) == 1:
				return matrix[0][0] 
			det = 0
			for j in range(len(matrix)):
				minor = [row[:j] + row[j+1:] for row in matrix[1:]]
				det += (-1)**(j) * matrix[0][j] * determinant(minor)
			return det	
		
		the_trace = sum(np.diag(matrix))
		the_det = determinant(matrix)
		return (float(the_det), float(the_trace))
		
```

***Calculate F1 Score from Predicted and True Labels***

```python
def calculate_f1_score(y_true, y_pred):
	"""
	Calculate the F1 score based on true and predicted labels.

	Args:
		y_true (list): True labels (ground truth).
		y_pred (list): Predicted labels.

	Returns:
		float: The F1 score rounded to three decimal places.
	"""
	# Your code here
	TP = sum( 1 for i , j in zip(y_true , y_pred) if i == 1 and j == 1)
	TN = sum( 1 for i , j in zip(y_true , y_pred) if i == 0 and j == 0 )
	FP = sum( 1 for i , j in zip(y_true , y_pred) if i == 0 and j == 1 )
	FN = sum( 1 for i , j in zip(y_true , y_pred) if i == 1 and j == 0 )
	
	Precision = 0 if (TP + FP) == 0 else TP / (TP + FP)
	Recall = 0 if (TP + FN) == 0 else TP / (TP + FN)

	f1 =  0 if (Precision + Recall) == 0 else 2 * ((Precision * Recall) / (Precision + Recall))
	return round(f1,3)

```

```python
# Precision-Recall Curve implementation

def precision_recall_curve(y_true: list, y_scores: list) -> tuple:
	 """
	 Compute precision-recall pairs for different probability thresholds.
	 
	 Args:
		  y_true: List of true binary labels (0 or 1)
		  y_scores: List of predicted probabilities or confidence scores
	 
	 Returns:
		  Tuple of (precisions, recalls, thresholds) where each is a list
	 """
	 # Your code here
	 precisions  = []
	 recalls = []
	 total_positives = sum(y_true)
	 thresholds = sorted(set(y_scores), reverse=True)
	 for threshold in thresholds:
		TP = 0
		FP = 0
		for score, label in zip(y_scores, y_true):
		  if score >= threshold:
				if label == 1:
					TP += 1
				else:
					FP += 1
		if TP + FP == 0:
			precision = 1
		else:
			precision = TP / (TP + FP)
		precisions.append(precision)
		if total_positives == 0:
			recall = 0.0
		else:
			recall = TP / total_positives
		recalls.append(recall)
	 
	 return precisions, recalls, thresholds  


```

```python
from typing import Tuple
def early_stopping(val_losses, patience, min_delta):
	best_loss = val_losses[0]
	best_epoch = 0
	epochs_without_improvement = 0

	for epoch in range(1, len(val_losses)):
		current_loss = val_losses[epoch]

		if best_loss - current_loss > min_delta:
			best_loss = current_loss
			best_epoch = epoch
			epochs_without_improvement = 0
		else:
			epochs_without_improvement += 1

		if epochs_without_improvement == patience:
			return (epoch, best_epoch)

	return (len(val_losses) - 1, best_epoch)


# Test case 1
losses1 = [0.9, 0.8, 0.75, 0.77, 0.76, 0.77, 0.78]
print(early_stopping(losses1, patience=2, min_delta=0.01))  # Expected: (4, 2)

# Test case 2
losses2 = [0.9, 0.8, 0.79, 0.78, 0.77]
print(early_stopping(losses2, patience=2, min_delta=0.1))   # Expected: (4, 2)
```

***Macro F1 Score for Multi-Class Classification***

```python
def macro_f1(y_true: list, y_pred: list) -> float:
	"""
	Compute the Macro F1 score over all classes appearing in y_true or y_pred.
	Returns a single float.
	"""
	total_F1 = []
	# input : y_true = [0, 0, 1, 1, 2, 2], y_pred = [0, 1, 1, 1, 2, 0]
	union = set(y_true) | set(y_pred) # union = {0, 1, 2} | {0, 1, 2} = {0, 1, 2}
	for c in union:
		TP = sum(1 for t, p in zip(y_true, y_pred) if (t == c and p == c))
		FP = sum(1 for t, p in zip(y_true, y_pred) if (t != c and p == c))
		FN = sum(1 for t, p in zip(y_true, y_pred) if (t == c and p != c))
		if TP + FP == 0 :
			Precision = 0
		else :
			Precision = TP / (TP + FP)
		if TP + FN == 0 :
			Recall = 0
		else :
			Recall = TP / (TP + FN)
		if Precision + Recall == 0 :
			F_1 = 0 
		else :
			F_1 = 2 * ((Precision * Recall) / (Precision + Recall))
		total_F1.append(F_1)
	macro_average = sum(total_F1) / len(union)
	return macro_average 

print(round(macro_f1([0, 1, 2, 0, 1, 2], [0, 1, 2, 0, 1, 2]), 4))
```

***Micro F1 Score for Multi-Label Classification***

```python
import numpy as np

def micro_f1_score(y_true: np.ndarray, y_pred: np.ndarray) -> float:
	
	TP = np.sum((y_true == 1) & (y_pred == 1))
	TN = np.sum((y_true == 0) & (y_pred == 0))
	FN = np.sum((y_true == 1) & (y_pred == 0))
	FP = np.sum((y_true == 0) & (y_pred == 1))

	if TP + FP != 0:
		P_micro = TP / (TP + FP)
	else:
		P_micro = 0

	if TP + FN != 0:
		R_micro = TP / (TP + FN)
	else:
		R_micro = 0

	if P_micro + R_micro != 0:
		F1_micro = 2 * (P_micro * R_micro) / (P_micro + R_micro)
	else:
		F1_micro = 0

	return F1_micro

y_true = np.array([[1,0,1],[0,1,0]])
y_pred = np.array([[1,0,1],[0,1,0]])
print(round(micro_f1_score(y_true, y_pred), 4))
```

***k-fold cross-validation***

```python
import numpy as np
from typing import List, Tuple

def k_fold_cross_validation(n_samples: int, k: int = 5, shuffle: bool = True) -> List[Tuple[List[int], List[int]]]:
	
	indices = np.arange(n_samples) 
	if shuffle :
		np.random.shuffle(indices)
	folds = []
	start = 0
	base_size = n_samples // k
	remainder = n_samples %  k 
	for i in range(k):
		if i < remainder:
			fold_size = base_size + 1
		else:
			fold_size = base_size 
		end = start + fold_size
		folds.append(indices[start:end])
		start = end
	result = []
	for i in range(k):
		test_indices  = folds[i]        # just fold i
		train_indices = np.concatenate([folds[j] for j in range(k) if j != i])         # all other folds combined
		result.append((train_indices, test_indices))
	return result
		
```

***Calculate AUC (Area Under ROC Curve)***

```python
import numpy as np

def calculate_auc(y_true, y_scores):
	
	if len(set(y_true)) == 1:
		return 0.0
	
	result = []
	thresholds = [float('inf')] + sorted(set(y_scores), reverse=True)
	N = sum([1 for i in y_true if i == 0])
	P = sum([1 for i in y_true if i == 1])
	for threshold in thresholds :
		TP = 0
		FP  = 0
		for i in range(len(y_scores)):
			if y_scores[i] >= threshold :
				if y_true[i] == 1 :
					TP += 1
				else :
					FP += 1
		TPR = TP / P
		FPR = FP / N
		result.append((TPR,FPR))
	
	tpr_list, fpr_list = zip(*result)
	auc = 0
	for i in range(1,len(fpr_list)):
		auc += ((tpr_list[i-1] + tpr_list[i]) / 2) * (fpr_list[i] - fpr_list[i-1])
	return auc
	
```

```python
def apply_weight_decay(parameters: list[list[float]], gradients: list[list[float]], lr: float, weight_decay: float, apply_to_all: list[bool]) -> list[list[float]]:

	for i in range(len(parameters)):
		if apply_to_all[i]: 
			for j in range(len(gradients[i])):
				# wnew​=w−η⋅g−η⋅λ⋅w 
				w_new = parameters[i][j] - (lr * gradients[i][j]) - (lr * weight_decay * parameters[i][j] )
				parameters[i][j] = w_new
		else :
			for j in range(len(gradients[i])): 
				# w - lr * g 
				w_new = parameters[i][j] - (lr * gradients[i][j])
				parameters[i][j] = w_new
	return parameters

print([[round(x, 4) for x in arr] for arr in apply_weight_decay([[10.0, -5.0]], [[1.0, -0.5]], 0.01, 0.1, [True])])
print([[round(x, 4) for x in arr] for arr in apply_weight_decay([[1.0, 2.0], [0.5, 0.5]], [[0.1, 0.2], [0.05, 0.05]], 0.1, 0.01, [True, False])])
```

```python
import numpy as np

def weighted_f1_score(y_true, y_pred) -> float:
	
	total_F1 = [] 
	union = set(y_true) | set(y_pred)  

	for n in union :
		TP = sum(1 for t, p in zip(y_true, y_pred) if (t == n and p == n))
		FP = sum(1 for t, p in zip(y_true, y_pred) if (t != n and p == n))
		FN = sum(1 for t, p in zip(y_true, y_pred) if (t == n and p != n))
		# This counts how many times class n appears in y_true
		support = sum(1 for t in y_true if t == n)
		
		if TP + FP == 0 :
			 Precision = 0 
		else :
			Precision = TP / (TP + FP)
		if TP + FN == 0 :
			Recall = 0 
		else :
			Recall = TP / (TP + FN)

		if Precision + Recall == 0 :
			f1 = 0 
		else :
			f1 = F_1 = 2 * ((Precision * Recall) / (Precision + Recall))
		# Instead of appending just f1, we multiply it by the support first
		total_F1.append(f1 * support)
		
	return round(sum(total_F1) / len(y_true), 4)

print(weighted_f1_score(np.array([0,1,2,0,1,2]), np.array([0,2,1,0,0,1])))
print(weighted_f1_score(np.array([0,1,2]), np.array([1,2,0])))
```

```python
import math

PI = 3.14159

def power_grid_forecast(consumption_data):
	# 1) Subtract the daily fluctuation (10 * sin(2π * i / 10)) from each data point.
	# 2) Perform linear regression on the detrended data.
	# 3) Predict day 15's base consumption.
	# 4) Add the day 15 fluctuation back.
	# 5) Round, then add a 5% safety margin (rounded up).
	# 6) Return the final integer.
	n = len(consumption_data)
	sum_x = 0  															# add i
	sum_y = 0  															#  add detrended
	sum_xy = 0  														# add i * detrended
	sum_x2 = 0  														#  add i * i
	for i in range(1,len(consumption_data) + 1):
		fluctuation = (10 * math.sin((2*PI*i) / 10))
		detrended = consumption_data[i-1] - fluctuation
		sum_x += i
		sum_y += detrended
		sum_xy += i * detrended
		sum_x2 += i * i

	m = ((n * sum_xy) - (sum_x * sum_y)) / ((n * sum_x2) - sum_x**2)
	b = (sum_y - m*sum_x) / n
	base_15 = m * 15 + b
	f_15 = 10 * math.sin(2 * PI * 15 / 10)
	pred_15 = base_15 + f_15
	final_15 = math.ceil(1.05 * round(pred_15))

	return final_15

print(power_grid_forecast([150, 165, 185, 195, 210, 225, 240, 260, 275, 290]))
print(power_grid_forecast([150, 165, 185, 195, 210, 225, 240, 260, 275, 290]))
print(power_grid_forecast([160, 170, 190, 200, 215, 230, 245, 265, 280, 295]))
```

***Implement Decision Tree for Regression***

```python
import numpy as np

def decision_tree_regressor(X_train, y_train, X_test, max_depth=2, min_samples_split=2):
	
	def mse(y): 
		MSE = np.mean((y - np.mean(y))**2)
		return MSE
	
	def find_best_split(X, y):
		best_gain = -np.inf
		best_feature = None
		best_threshold = None
		for i in range(X.shape[1]):
			  unique_vals = np.unique(X[:, i])
			  thresholds = (unique_vals[:-1] + unique_vals[1:]) / 2 
			  for threshold in thresholds :
				  left_mask = X[:, i] <= threshold
				  right_mask = X[:, i] > threshold
				  y_left = y[left_mask]
				  y_right = y[right_mask]
				  if len(y_left) == 0 or len(y_right) == 0:
					  continue 
				  weighted_MSE = (len(y_left)*mse(y_left) + len(y_right)*mse(y_right)) / (len(y_right)+len(y_left))
				  Gain = mse(y) - weighted_MSE
				  if Gain > best_gain:
					  best_gain = Gain
					  best_feature = i
					  best_threshold = threshold
		return best_feature, best_threshold
					
	def build_tree(X, y, depth): 
		if depth >= max_depth or len(X)< min_samples_split or len(np.unique(y)) == 1:
			return {'leaf': True, 'value': np.mean(y)}
		feature, threshold = find_best_split(X, y) 
		if feature is None: 
			return {'leaf': True, 'value': np.mean(y)}
		
		left_mask = X[:, feature] <= threshold
		right_mask = X[:, feature] > threshold
		y_left = y[left_mask]
		y_right = y[right_mask]
		X_left = X[left_mask]
		X_right = X[right_mask]
		left_subtree = build_tree(X_left, y_left, depth + 1)
		right_subtree = build_tree(X_right, y_right, depth + 1)
		return {'leaf': False, 'feature': feature, 'threshold': threshold, 'left': left_subtree, 'right': right_subtree}
	
	def predict_sample(node, x) :
		if node['leaf'] == True :
			return node['value']
		elif x[node['feature']] <= node['threshold']  :
			return predict_sample(node['left'], x)
		else :
			return predict_sample(node['right'], x)
	X_train = np.array(X_train)
	y_train = np.array(y_train)
	X_test = np.array(X_test)
	tree = build_tree(X_train, y_train, 0)
	predictions = []
	for sample in X_test :
		predictions.append(predict_sample(tree, sample))
	return np.round(predictions, 4).tolist()

print(decision_tree_regressor([[0], [1], [2], [3]], [0.0, 0.0, 1.0, 1.0], [[0.5], [2.5]], max_depth=1, min_samples_split=2))
print(decision_tree_regressor([[0], [1], [2], [3]], [1.0, 2.0, 3.0, 4.0], [[1], [2]], max_depth=0, min_samples_split=2))
```

***Decision Tree Learning***

```python
import math
from collections import Counter
import numpy as np


def calculate_entropy(labels: list) -> float:
	"""Calculate the entropy of a list of labels."""
	counts = Counter(labels) 
	total = len(labels)
	entropy = 0 
	for count in counts.values():
		p = count / total
		entropy -= p * math.log2(p)
	return entropy

def calculate_information_gain(examples: list[dict], attr: str, target_attr: str) -> float:
	"""Calculate the information gain of splitting on attr."""
	labels = [example[target_attr] for example in examples]
	parent_entropy = calculate_entropy(labels)
	attr_values = [example[attr] for example in examples]
	unique_values = set(attr_values)
	weighted_entropy = 0
	for value in sorted(unique_values):
		subset = [example for example in examples if example[attr] == value]
		weight = len(subset) / len(examples)  
		subset_labels = [example[target_attr] for example in subset]
		weighted_entropy += weight * calculate_entropy(subset_labels)
	return parent_entropy - weighted_entropy

def majority_class(examples: list[dict], target_attr: str) -> str:
	"""Return the majority class. Break ties alphabetically."""
	labels = [example[target_attr] for example in examples]
	counts = Counter(labels)
	# Get all classes that have the max count
	max_count = max(counts.values())
	# Filter to only classes with that count using a list comprehension
	top_classes = [cls for cls, count in counts.items() if count == max_count]
	# Pick the alphabetically first one using min(), which works alphabetically on strings
	return min(top_classes)

def learn_decision_tree(examples: list[dict], attributes: list[str], target_attr: str) -> dict:
	"""Build a decision tree using the ID3 algorithm."""

	labels = [example[target_attr] for example in examples]
	counts = Counter(labels)
	if len(examples) == 0 or len(attributes) == 0:
	 return majority_class(examples, target_attr)
	if len(counts) == 1 :  # This says: "if there's only one unique class, return it"
	 return labels[0]  
	best_attr = max(attributes, key=lambda attr: calculate_information_gain(examples, attr, target_attr))  # IG for each one and pick the best one
	remaining_attrs = [attr for attr in attributes if attr != best_attr]
	tree = {best_attr: {}} 
	unique_values = sorted(set(example[best_attr] for example in examples))
	for value in unique_values :
	   subset = [example for example in examples if example[best_attr] == value]
	   tree[best_attr][value] = learn_decision_tree(subset, remaining_attrs, target_attr)
	return tree


print(learn_decision_tree([
	{'Outlook': 'Sunny', 'Wind': 'Weak', 'PlayTennis': 'No'},
	{'Outlook': 'Overcast', 'Wind': 'Strong', 'PlayTennis': 'Yes'},
	{'Outlook': 'Rain', 'Wind': 'Weak', 'PlayTennis': 'Yes'},
	{'Outlook': 'Sunny', 'Wind': 'Strong', 'PlayTennis': 'No'},
	{'Outlook': 'Overcast', 'Wind': 'Weak', 'PlayTennis': 'Yes'},
	{'Outlook': 'Rain', 'Wind': 'Strong', 'PlayTennis': 'No'},
	{'Outlook': 'Rain', 'Wind': 'Weak', 'PlayTennis': 'Yes'}
], ['Outlook', 'Wind'], 'PlayTennis'))

print(learn_decision_tree([
	{'Outlook': 'Sunny', 'Wind': 'Weak', 'PlayTennis': 'No'},
	{'Outlook': 'Overcast', 'Wind': 'Strong', 'PlayTennis': 'Yes'},
	{'Outlook': 'Rain', 'Wind': 'Weak', 'PlayTennis': 'Yes'},
	{'Outlook': 'Sunny', 'Wind': 'Strong', 'PlayTennis': 'No'},
	{'Outlook': 'Overcast', 'Wind': 'Weak', 'PlayTennis': 'Yes'},
	{'Outlook': 'Rain', 'Wind': 'Strong', 'PlayTennis': 'No'},
	{'Outlook': 'Rain', 'Wind': 'Weak', 'PlayTennis': 'Yes'}
], ['Outlook', 'Wind'], 'PlayTennis'))
```

```python
import  numpy as np

def binary_cross_entropy(y_true: list[float], y_pred: list[float], epsilon: float = 1e-15) -> float:
	"""
	Compute binary cross-entropy loss.
	
	Args:
		y_true: True binary labels (0 or 1)
		y_pred: Predicted probabilities (between 0 and 1)
		epsilon: Small value for numerical stability
	
	Returns:
		Mean binary cross-entropy loss
	"""
	# Your code
	y_pred = np.array(y_pred)
	y_true = np.array(y_true) 
	y_pred_clipped = np.clip(y_pred, epsilon, 1 - epsilon)
	loss = y_true * np.log(y_pred_clipped) + (1 - y_true)*np.log(1- y_pred_clipped)
	return - np.mean(loss)


result = binary_cross_entropy([1, 0, 1, 0], [0.5, 0.5, 0.5, 0.5])
print(round(result, 4))
print("-------")
result = binary_cross_entropy([1, 1, 1, 0, 0, 0], [0.99, 0.95, 0.85, 0.15, 0.05, 0.01])
print(round(result, 4))

```

```python
import numpy as np

def sigmoid(z):
	return 1 / (1 + np.exp(-z))

def train_neuron(features, labels, initial_weights, initial_bias, learning_rate, epochs):
	features = np.array(features)
	labels = np.array(labels)
	w = np.array(initial_weights, dtype=float).copy()
	b = initial_bias
	mse_values = []

	for epoch in range(epochs):
		z = np.dot(features, w) + b
		y_hat = sigmoid(z)
		MSE = np.mean((y_hat - labels)**2)
		delta = (y_hat - labels) * y_hat * (1 - y_hat)
		dL_dw = 2 * np.dot(features.T, delta) / len(features)
		dL_db = 2 * np.mean(delta)

		w = w - learning_rate * dL_dw
		b = b - learning_rate * dL_db

		mse_values.append(round(MSE, 4))

	return np.round(w, 4), round(b, 4), mse_values

print(train_neuron(np.array([[1.0, 2.0], [2.0, 1.0], [-1.0, -2.0]]), np.array([1, 0, 0]), np.array([0.1, -0.2]), 0.0, 0.1, 2))
```

***Character-Level Tokenizer (stoi/itos/BOS)***

```python
class CharTokenizer:
	def __init__(self, text: str):
		"""
		Build a character-level tokenizer from the input text.
		
		Args:
			text: A string used to build the vocabulary.
		"""
		# Your code here
		special = ['<BOS>', '<EOS>']
		chars = sorted(set(text))
		vocab = special + chars
		self.stoi = {ch: i for i, ch in enumerate(vocab)}   # char → index
		self.itos = {i: ch for i, ch in enumerate(vocab)}   # index → char
		self.vocab_size = len(vocab)

	def encode(self, text: str) -> list:
		"""
		Encode a string into a list of token indices.
		
		Args:
			text: The string to encode.
		Returns:
			List of integer indices.
		"""
		return [self.stoi['<BOS>']] + [self.stoi[ch] for ch in text] + [self.stoi['<EOS>']]
		
	def decode(self, indices: list) -> str:
		"""
		Decode a list of token indices back into a string.
		
		Args:
			indices: List of integer indices.
		Returns:
			Decoded string.
		"""
		# Your code here
		return "".join([self.itos[i] for i in indices])
	


tok = CharTokenizer('hello')
print(tok.vocab_size)

print("----------------")

tok = CharTokenizer('the quick brown fox')
print(tok.vocab_size)
encoded = tok.encode('fox')
print(encoded)
print(tok.decode(encoded))
```

```python
class Value:
	def __init__(self, data, _children=(), _op=''):
		self.data = data
		self.grad = 0
		self._backward = lambda: None
		self._prev = set(_children)
		self._op = _op
	def __repr__(self):
		def fmt(x):
			return int(x) if float(x).is_integer() else round(x, 4)
		return f"Value(data={fmt(self.data)}, grad={fmt(self.grad)})"

	def __add__(self, other):
		 # Implement addition here
		out = Value(self.data +  other.data,(self,other), '+')

		def _backward():
			self.grad += 1.0 * out.grad
			other.grad += 1.0 * out.grad
		out._backward = _backward
		return out 
	

	def __mul__(self, other):
		# Implement multiplication here
		out = Value(self.data *  other.data,(self,other), '*')

		def _backward():
			self.grad += other.data * out.grad
			other.grad += self.data * out.grad
		out._backward = _backward
		return out 

	def relu(self):
		out = Value(max(0, self.data), (self,), 'relu')
		def _backward():
			self.grad += (out.data > 0) * out.grad
		out._backward = _backward
		return out

	def backward(self):
		topo = []
		visited = set()
		def build_topo(v):
			if v not in visited:
				visited.add(v)
				for child in v._prev:
					build_topo(child)
				topo.append(v)
		build_topo(self)
		self.grad = 1.0
		for node in reversed(topo):
			node._backward()


a = Value(2);b = Value(3);c = Value(10);d = a + b * c  ;e = Value(7) * Value(2);f = e + d;g = f.relu()  
g.backward()
print(a,b,c,d,e,f,g)
print("-----------------------------------")
a = Value(2); b = Value(3); m = a * b; y = m + a; z = m + b; w = y + z
w.backward()
print(a, b, m, y, z, w)
```

***Average Negative Log-Likelihood for Bigram Model***

```python
import numpy as np

def bigram_avg_nll(P, words, stoi):
	"""Average negative log-likelihood of words under a bigram model."""
	N = 0
	log_likelihood  = 0.0
	for word in words :
		
		chs = ["."] + list((word)) + ['.']
		for ch1, ch2 in zip(chs, chs[1:]):
		   p = P[stoi[ch1], stoi[ch2]]  
		   log_likelihood += np.sum(np.log(p))
		   N += 1
	if N == 0:
		return 0.0
	return round(-log_likelihood / N, 4)

stoi = {'.':0,'a':1,'b':2}
P = np.array([[0.1,0.6,0.3],[0.4,0.2,0.4],[0.5,0.3,0.2]])
print(bigram_avg_nll(P, ['ab'], stoi))
print("-----------------------------------------------------")
stoi = {'.':0,'a':1,'b':2,'c':3}
P = np.array([[0.25,0.25,0.25,0.25],[0.1,0.2,0.3,0.4],[0.4,0.3,0.2,0.1],[0.25,0.25,0.25,0.25]])
print(bigram_avg_nll(P, ['abc','ca'], stoi))
```

```python
import numpy as np

def smooth_bigram_probs(N, k):
	"""
	N: 2D list or array of bigram counts (V x V)
	k: float, add-k smoothing constant (k >= 0)
	Returns: 2D list of smoothed bigram probabilities (V x V).
	"""
	N = np.array(N)
	smoothed = N + k
	row_sums = np.sum(smoothed, axis=1, keepdims=True)
	P = smoothed / row_sums
	return  P.tolist()

N = [[2, 0], [1, 3]]
P = smooth_bigram_probs(N, 1)
print([[round(x, 5) for x in row] for row in P])
print("------------------------------")
N = [[0, 0, 0], [5, 0, 5], [1, 2, 3]]
P = smooth_bigram_probs(N, 1)
print([[round(x, 5) for x in row] for row in P])
```

***Partial Derivatives of Multivariable Functions***

```python
import numpy as np

def compute_partial_derivatives(func_name: str, point: tuple[float, ...]) -> tuple[float, ...]:

	# unpack the point ONCE, at the top
	if len(point) == 2:
		x, y = point
	else:
		x, y, z = point

	if func_name == 'poly2d':
		df_dx = 2*x*y + y**2
		df_dy = x**2 + 2*x*y
		return (df_dx, df_dy)

	elif func_name == 'exp_sum':
		df_dx = np.exp(x+y)
		df_dy = np.exp(x+y)
		return (df_dx, df_dy)

	elif func_name == 'product_sin':
		df_dx = np.sin(y)
		df_dy = x*np.cos(y)
		return (df_dx, df_dy)

	elif func_name == 'poly3d':
		df_dx = 2*x*y
		df_dy = x**2 + z**2
		df_dz = 2*y*z
		return (df_dx, df_dy, df_dz)

	elif func_name == 'squared_error':
		df_dx =  2*(x-y)      
		df_dy = -2*(x-y)
		return (df_dx, df_dy)
	
result = compute_partial_derivatives('poly2d', (2.0, 3.0)); print(f"{result[0]:.1f},{result[1]:.1f}")
print("-----------------------------------")
result = compute_partial_derivatives('poly3d', (1.0, 2.0, 3.0)); print(f"{result[0]:.1f},{result[1]:.1f},{result[2]:.1f}")
```

***Implement the Tanh Activation Function***

```python
import math

def tanh(x: float) -> float:
	"""
	Implements the Tanh (hyperbolic tangent) activation function.

	Args:
		x (float): Input value

	Returns:
		float: The tanh of the input, rounded to 4 decimal places
	"""
	# Your code here
	return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))

print(tanh(-1)) 
print('--------------')
print(tanh(2))
print('--------------')
print(tanh(-2))
```

***Calculate Number of Parameters in Neural Network***

```python
def calculate_parameters(layers: list[dict]) -> int:
	total = 0
	for layer in layers:

		if layer['type'] == 'dense':
			w = layer['input_size'] * layer['output_size']
			if layer.get('bias', True):        # missing key → defaults to True
				w += layer['output_size']
			total += w

		elif layer['type'] == 'conv2d':
			w = layer['in_channels'] * layer['out_channels'] * layer['kernel_size']**2
			if layer.get('bias', True):
				w += layer['out_channels']
			total += w

	return total

print(calculate_parameters([{'type': 'conv2d', 'in_channels': 1, 'out_channels': 8, 'kernel_size': 5, 'bias': True}, #layer 1
							{'type': 'dense', 'input_size': 128, 'output_size': 64, 'bias': False}, # layer 2
							{'type': 'dense', 'input_size': 64, 'output_size': 10, 'bias': True}]))
```

***PyTorch: Build a Complete Training Loop***



```python
import torch
import torch.nn as nn
import torch.optim as optim

def train_model(model, X_train, y_train, X_val, y_val, epochs, batch_size, lr):
	history = []

	# setup
	optimizer = optim.Adam(model.parameters(), lr=lr)
	criterion = nn.CrossEntropyLoss()
	n_samples = X_train.shape[0]

	for epoch in range(1, epochs + 1):          # 1-indexed, as the spec wants

		# ---- train ----
		model.train()

		# shuffle (X and y with the SAME perm — flashcards stay married)
		perm = torch.randperm(n_samples)
		X_shuffled = X_train[perm]
		y_shuffled = y_train[perm]

		epoch_loss = 0.0
		n_batches = 0

		# batch loop
		for i in range(0, n_samples, batch_size):
			X_batch = X_shuffled[i : i + batch_size]
			y_batch = y_shuffled[i : i + batch_size]

			# the training step — yours, unchanged
			optimizer.zero_grad()
			outputs = model(X_batch)
			loss = criterion(outputs, y_batch)
			loss.backward()
			optimizer.step()

			epoch_loss += loss.item()           # .item() → plain float, no memory leak
			n_batches += 1

		# ---- validate ----
		model.eval()
		with torch.no_grad():
			val_logits = model(X_val)
			val_loss = criterion(val_logits, y_val).item()
			preds = val_logits.argmax(dim=1)
			val_accuracy = (preds == y_val).float().mean().item()

		# ---- record ----
		history.append({
			'epoch': epoch,
			'train_loss': epoch_loss / n_batches,
			'val_loss': val_loss,
			'val_accuracy': val_accuracy,
		})

	return history
```

***MNIST: Build Neural Network from Scratch (NumPy Only)***


```python
import numpy as np
from sklearn.datasets import fetch_openml


class NeuralNetwork:
	'''
	Neural network from scratch using only NumPy.
	Architecture: 784 → 128 (ReLU) → 10 (Softmax)
	'''
	def __init__(self, input_size=784, hidden_size=128, output_size=10, lr=0.01):
		self.lr = lr
		self.W1 = np.random.randn(input_size, hidden_size) * np.sqrt(2 / input_size)
		self.b1 = np.zeros((1, hidden_size))
		self.W2 = np.random.randn(hidden_size, output_size) * np.sqrt(2 / hidden_size)
		self.b2 = np.zeros((1, output_size))

	def forward(self, X):
		z1 = X @ self.W1 + self.b1
		h1 = np.maximum(z1, 0)                                  # ReLU
		z2 = h1 @ self.W2 + self.b2
		z_shifted = z2 - np.max(z2, axis=1, keepdims=True)      # numerical stability
		exp_z = np.exp(z_shifted)
		probs = exp_z / np.sum(exp_z, axis=1, keepdims=True)    # softmax, (N, 10)
		self.cache = {'X': X, 'z1': z1, 'h1': h1, 'z2': z2}
		return probs

	def backward(self, X, y, probs):
		N = X.shape[0]

		# 1. Cross-entropy loss
		correct_probs = probs[np.arange(N), y]
		loss = -np.mean(np.log(correct_probs + 1e-12))

		# 2. Gradient at output (softmax + cross-entropy combined)
		dz2 = probs.copy()
		dz2[np.arange(N), y] -= 1
		dz2 /= N

		# 3. Backprop through layer 2
		h1 = self.cache['h1']
		dW2 = h1.T @ dz2
		db2 = np.sum(dz2, axis=0, keepdims=True)
		dh1 = dz2 @ self.W2.T

		# 4. Backprop through ReLU (kill gradient where z1 <= 0)
		dz1 = dh1 * (self.cache['z1'] > 0)

		# 5. Backprop through layer 1
		dW1 = X.T @ dz1
		db1 = np.sum(dz1, axis=0, keepdims=True)

		# 6. Update all parameters
		self.W2 -= self.lr * dW2
		self.b2 -= self.lr * db2
		self.W1 -= self.lr * dW1
		self.b1 -= self.lr * db1

		return loss

	def train_step(self, X, y):
		probs = self.forward(X)
		loss = self.backward(X, y, probs)
		return loss

	def predict(self, X):
		probs = self.forward(X)
		return np.argmax(probs, axis=1)


# ---------- Load MNIST ----------
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = mnist.data.astype(np.float64) / 255.0   # scale pixels 0-255 → 0-1
y = mnist.target.astype(np.int64)

X_train, X_test = X[:60000], X[60000:]
y_train, y_test = y[:60000], y[60000:]

# ---------- Training loop ----------
model = NeuralNetwork(lr=0.1)
batch_size = 64

for epoch in range(5):
	# shuffle each epoch
	perm = np.random.permutation(len(X_train))
	X_train, y_train = X_train[perm], y_train[perm]

	# mini-batch training
	for i in range(0, len(X_train), batch_size):
		X_batch = X_train[i:i+batch_size]
		y_batch = y_train[i:i+batch_size]
		loss = model.train_step(X_batch, y_batch)

	# evaluate on unseen test data
	acc = np.mean(model.predict(X_test) == y_test)
	print(f"epoch {epoch+1}: test accuracy = {acc:.4f}")
```

***Implement the ELU Activation Function***

```python
import numpy as np

def elu(x: float, alpha: float = 1.0) -> float:
	"""
	Compute the ELU activation function.

	Args:
		x (float): Input value
		alpha (float): ELU parameter for negative values (default: 1.0)

	Returns:
		float: ELU activation value
	"""
	# Your code here
	if x > 0:
		val = x
	else:
		val = alpha*(np.exp(x) - 1)
		
	return round(val,4)

print(elu(-1))
print(elu(-1, alpha=2.0))
```

***Compute TPR and FPR from Classifications***

```python
import numpy as np

def compute_tpr_fpr(y_true, y_pred):
	"""
	Compute TPR and FPR from true and predicted binary labels.

	Args:
		y_true (array-like): Ground-truth labels (0 or 1).
		y_pred (array-like): Predicted labels (0 or 1).

	Returns:
		tuple: (tpr, fpr) as Python floats.
	"""
	# TODO: compute TP, FN, FP, TN then TPR and FPR
	TP = np.sum((y_true == 1) & (y_pred == 1))
	TN = np.sum((y_true == 0) & (y_pred == 0))
	FN = np.sum((y_true == 1) & (y_pred == 0))
	FP = np.sum((y_true == 0) & (y_pred == 1))

	if TP + FN == 0 :
		TPR = 0.0
	else :
		TPR = TP / (TP + FN)

	if FP + TN == 0 :
		FPR = 0.0 
	else :
		FPR = FP / (FP + TN)
	return (float(TPR), float(FPR))

print(compute_tpr_fpr(np.array([1, 1, 0, 0]), np.array([1, 0, 1, 0])))
```
