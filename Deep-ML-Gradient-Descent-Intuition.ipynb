***Single Neuron***

```python
import math
import numpy as np 


def single_neuron_model(features: list[list[float]], labels: list[int], weights: list[float], bias: float) :
	# Your code here
	z=np.dot(features,weights)+bias
	sigma = 1 / (1 + np.exp(-z))   # sigmoid activation 
	mse = np.mean((sigma - labels)**2)
	
	return np.round(sigma,4).tolist(), np.round(mse,4).tolist()

print(single_neuron_model([[0.5, 1.0], [-1.5, -2.0], [2.0, 1.5]], [0, 1, 0], [0.7, -0.4], -0.1))
```

***Sigmoid Activation Function Understanding***

```python
import math

def sigmoid(z: float) -> float:
	#Your code here
	result = 1 / (1+math.exp(-z))
	return result

print(sigmoid(0))
```

***Calculate Number of Parameters in Neural Network***

```python
def calculate_parameters(layers: list[dict]) -> int:
	total = 0
	for layer in layers:

		if layer['type'] == 'dense':
			w = layer['input_size'] * layer['output_size']
			if layer.get('bias', True):        # missing key → defaults to True
				w += layer['output_size']
			total += w

		elif layer['type'] == 'conv2d':
			w = layer['in_channels'] * layer['out_channels'] * layer['kernel_size']**2
			if layer.get('bias', True):
				w += layer['out_channels']
			total += w

	return total

print(calculate_parameters([{'type': 'dense', 'input_size': 784, 'output_size': 128, 'bias': True}, 
							{'type': 'dense', 'input_size': 128, 'output_size': 10, 'bias': True}]))
```

***Implement the Huber Loss Function***

```python
import numpy as np 

def huber_loss(y_true, y_pred, delta=1.0):
	
	y_true = np.array(y_true)
	y_pred = np.array(y_pred)

	abs_error = np.abs(y_true - y_pred)

    #  np.where(condition, value_if_true, value_if_false)
	loss = np.where(
	abs_error <= delta,
	0.5 * abs_error**2,
	delta * (abs_error - 0.5 * delta))

	return np.mean(loss)

print(round(huber_loss([2.0, 3.0, 4.0], [2.5, 2.0, 4.0], delta=1.0), 4))
print(round(huber_loss([0.0, 0.0, 0.0], [2.0, 0.5, 1.0], delta=1.0), 4))
```

```python
import numpy as np

def huber_loss(y_true, y_pred, delta=1.0):

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    losses = []

    for yt, yp in zip(y_true, y_pred):
        error = abs(yt - yp)

        if error <= delta:
            loss = 0.5 * error**2
        else:
            loss = delta * (error - 0.5 * delta)

        losses.append(loss)

    return float(np.mean(losses))

print(round(huber_loss([2.0, 3.0, 4.0], [2.5, 2.0, 4.0], delta=1.0), 4))
print(round(huber_loss([0.0, 0.0, 0.0], [2.0, 0.5, 1.0], delta=1.0), 4))
```

***Compute Multi-class Cross-Entropy Loss***

```python
import numpy as np

def compute_cross_entropy_loss(predicted_probs, true_labels, epsilon=1e-15):

    predicted_probs = np.clip(predicted_probs, epsilon, 1 - epsilon)

    loss = -np.sum(true_labels * np.log(predicted_probs), axis=1)

    return float(np.mean(loss))

pred = np.array([[1, 0, 0], [0, 1, 0]])
true = np.array([[1, 0, 0], [0, 1, 0]])
print(round(compute_cross_entropy_loss(pred, true), 4))
```
