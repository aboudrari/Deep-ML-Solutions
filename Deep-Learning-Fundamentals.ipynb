***Sigmoid Activation Function Understanding***

```python
# Sigmoid Activation Function
# σ(x) = 1 / (1 + e^(-x))
# Squeezes any number into range (0, 1)
# Used for: binary classification (output = probability)

import math

def sigmoid(z: float) -> float:
	#Your code here
	result = 1 / (1+math.exp(-z))
	return result

print(sigmoid(0))
print(sigmoid(1))
print(sigmoid(-1))
```

```python
# Using PyTorch
# "If x is large → sigmoid ≈ 1, if x is zero → sigmoid = 0.5, if x is very negative → sigmoid ≈ 0"
import torch

def sigmoid(x: float) -> float:
    t = torch.tensor(x)
    result = 1 / (1 + torch.exp(-t))
    return result.item()

print(sigmoid(0))
print(sigmoid(1))
print(sigmoid(-1))
```

***Implement ReLU Activation Function***

```python
# "Negative? → 0. Positive? → keep it as is." 
def relu(x: float) -> float:
    return max(0.0, x)

print(relu(0.0))
print(relu(1.0))
print(relu(-1.0))
```

```python
import torch 
def relu(z: float) -> float:
	# Your code here
	return torch.tensor(z).clamp(min=0).item()

print(relu(0.0))
print(relu(1.0))
print(relu(-1.0))
```

```python
import torch
import torch.nn.functional as F

def relu(x: float) -> float:
    t = torch.tensor(x)
    result = F.relu(t)
    return result.item()

print(relu(0))   
print(relu(1))  
print(relu(-1)) 
```
