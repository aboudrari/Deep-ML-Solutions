***Sigmoid Activation Function Understanding***

```python
# Sigmoid Activation Function
# σ(x) = 1 / (1 + e^(-x))
# Squeezes any number into range (0, 1)
# Used for: binary classification (output = probability)

import math

def sigmoid(z: float) -> float:
	#Your code here
	result = 1 / (1+math.exp(-z))
	return result

print(sigmoid(0))
print(sigmoid(1))
print(sigmoid(-1))
```

```python
# Using PyTorch
# "If x is large → sigmoid ≈ 1, if x is zero → sigmoid = 0.5, if x is very negative → sigmoid ≈ 0"
import torch

def sigmoid(x: float) -> float:
	t = torch.tensor(x)
	result = 1 / (1 + torch.exp(-t))
	return result.item()

print(sigmoid(0))
print(sigmoid(1))
print(sigmoid(-1))
```

***Implement ReLU Activation Function***

```python
# "Negative? → 0. Positive? → keep it as is." 
def relu(x: float) -> float:
	return max(0.0, x)

print(relu(0.0))
print(relu(1.0))
print(relu(-1.0))
```

***Leaky ReLU Activation Function***

```python
def leaky_relu(z: float, alpha: float = 0.01) -> float:
	return z if z > 0 else alpha * z

print(leaky_relu(5))
print(leaky_relu(1))
print(leaky_relu(-1))
```

```python
import torch

def leaky_relu(z: float, alpha: float = 0.01) -> float|int:
	z = torch.tensor(z)
	result = torch.where(z > 0,z , z * alpha)
	return result.item()

print(leaky_relu(5))
print(leaky_relu(1))
print(leaky_relu(-1))
```

```python
import torch    
import torch.nn.functional as F

def leaky_relu(z: float, alpha: float = 0.01) -> float:
	t = torch.tensor(z, dtype=torch.float32)
	return F.leaky_relu(t, negative_slope=alpha).item()

print(leaky_relu(5))
print(leaky_relu(1))
print(leaky_relu(-1))
```

***Softmax Activation Function Implementation***

```python
# "The sum is always 1, and each value represents how confident the model is about each class"

import math
import numpy as np 
def softmax(scores):
	maxx = np.max(scores)     
	denom = np.sum([math.exp(z - maxx) for z in scores])       # compute once, using ALL scores
	
	results = []
	for score in scores:
		result = float(math.exp(score - maxx) / denom )   # now use maxx and denom here
		results.append(result)
	return results

print([round(x, 4) for x in softmax([1, 2, 3])])
```

```python
import torch

def softmax(scores: list) -> list:
	t = torch.tensor(scores, dtype=torch.float32)
	t = t - torch.max(t)          # step 1: subtract max
	exp = torch.exp(t) # step 2: exponentiate
	result = exp / torch.sum(exp)   # step 3: divide by sum
	return result.tolist()

print([round(x, 4) for x in softmax([1, 2, 3])])
```

***Implementation of Log Softmax Function***

```python
import numpy as np

def log_softmax(scores: list) -> np.ndarray:
	
	log_softmax_i = scores - np.max(scores) - np.log(np.sum(np.exp(scores - np.max(scores))))
	return log_softmax_i	

print(np.round(log_softmax([1, 2, 3]), 4))
```

```python
import torch

def log_softmax(scores: list) -> list:
	t = torch.tensor(scores, dtype=torch.float32)
	result =  (t - torch.max(t) - torch.log(torch.sum(torch.exp(t - torch.max(t))))).tolist()
	return [round(x, 4) for x in result]

print(log_softmax([1, 2, 3]))
```

***Compute Multi-class Cross-Entropy Loss***

```python
import numpy as np

def compute_cross_entropy_loss(predicted_probs: np.ndarray, true_labels: np.ndarray, epsilon = 1e-15) -> float:
	ones_ = []
	for probs_row, label_row in zip(predicted_probs, true_labels):
		position = np.argmax(label_row)           
		ones_.append(probs_row[position])     

	# 2. Clip for numerical stability (avoid log(0) = -inf)
	p = np.clip(np.array(ones_), epsilon, 1 - epsilon)

	# 3. Loss = -mean(log(p_correct))
	return -np.mean(np.log(p))

pred = np.array([[1, 0, 0], [0, 1, 0]])
true = np.array([[1, 0, 0], [0, 1, 0]])
print(round(compute_cross_entropy_loss(pred, true), 4))
```

```python
import numpy as np

def cross_entropy_loss(predicted_probs, true_labels):
	epsilon = 1e-15
	predicted_probs = np.clip(predicted_probs, epsilon, 1)      # clip between epsilon and 1 why : cuz --> log(0) = -infinity
	log_probs = np.log(predicted_probs)                      
	loss_per_sample = -np.sum(true_labels * log_probs, axis=1)  
	return round(np.mean(loss_per_sample), 4)                

pred = np.array([[1, 0, 0], [0, 1, 0]])
true = np.array([[1, 0, 0], [0, 1, 0]])
print(round(compute_cross_entropy_loss(pred, true), 4))
```

***Derivatives of Activation Functions***

```python
import numpy as np 
import math 

def activation_derivatives(x: float) -> dict[str, float]:
	
	# {'sigmoid': 0.25, 'tanh': 1.0, 'relu': 0.0} 
	def sigmoid_dev(x):
		# 
		sigmoid = (1 / (1 + np.exp(-x)))
		return float(sigmoid * (1 - sigmoid))
	
	def tanh_dev(x):
		tanh = ((math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x)))
		return float(1 - tanh**2)
	
	def relu_dev(x):
		if x > 0 :
			return 1.0
		else:
			return 0.0
	
	return {'sigmoid': sigmoid_dev(x), 'tanh': tanh_dev(x), 'relu': relu_dev(x)}

result = activation_derivatives(2.0)
print({k: round(v, 4) for k, v in result.items()})
result = activation_derivatives(-1.0)
print({k: round(v, 4) for k, v in result.items()})
```

```python
import torch

def activation_derivatives(x: float) -> dict:
	
	t = torch.tensor(x, requires_grad=True)  # requires_grad=True it means track the derivative of this one we will need it to bachward
	# Sigmoid
	sigmoid = torch.sigmoid(t)
	sigmoid.backward()
	sigmoid_grad = t.grad.item()
	
	# Tanh
	t = torch.tensor(x, requires_grad=True)  # reset
	tanh = torch.tanh(t)
	tanh.backward()
	tanh_grad = t.grad.item()
	
	# ReLU
	t = torch.tensor(x, requires_grad=True)  # reset
	relu = torch.relu(t)
	relu.backward()
	relu_grad = t.grad.item()
	
	return {'sigmoid': round(sigmoid_grad, 4), 
			'tanh': round(tanh_grad, 4), 
			'relu': round(relu_grad, 4)}

result = activation_derivatives(2.0)
print({k: round(v, 4) for k, v in result.items()})
result = activation_derivatives(-1.0)
print({k: round(v, 4) for k, v in result.items()})
```

***Derivative of Softmax***

```python
import  numpy as np

def softmax_derivative(x: list[float]) -> list[list[float]]:
	
	x = np.array(x)
	s = np.exp(x) / np.sum(np.exp(x))
	J = np.zeros((len(x), len(x)))
	for i in range(len(x)):
		for j in range(len(x)):
			if i == j :
				J[i][j] = s[i] * (1 - s[i])
			elif i != j :
				J[i][j] = -s[i]*s[j]
	return J

result = softmax_derivative([2.0, 1.0])
print([[round(v, 4) for v in row] for row in result])
result = softmax_derivative([0.0, 1.0])
print([[round(v, 4) for v in row] for row in result])
```

```python
import numpy as np

def softmax_derivative(x: list[float]) -> list[list[float]]:
    x = np.array(x)
    s = np.exp(x) / np.sum(np.exp(x))
    return np.diag(s) - np.outer(s, s)

result = softmax_derivative([2.0, 1.0])
print([[round(v, 4) for v in row] for row in result])
result = softmax_derivative([0.0, 1.0])
print([[round(v, 4) for v in row] for row in result])
```

***Implement a Linear Layer Forward Pass with Matrix Multiplication***

```python
import torch

def linear_forward(x: torch.Tensor, W: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    # TODO: implement y = x W^T + b using PyTorch ops
    # x.shape = (N, in_features) 
    # W.shape = (out_features, in_features)
    # b.shape = (out_features,) 
    W = torch.transpose(W, 0, 1)
    return torch.matmul(x, W) + b 

x = torch.tensor([[1.0, 2.0]])
W = torch.tensor([[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]])
b = torch.tensor([0.0, 0.0, 0.0])
print(linear_forward(x, W, b).tolist())
```

```python
import torch

def linear_forward(x: torch.Tensor, W: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    return x @ W.T + b

x = torch.tensor([[1.0, 2.0]])
W = torch.tensor([[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]])
b = torch.tensor([0.0, 0.0, 0.0])
print(linear_forward(x, W, b).tolist())
```

***Compute a Gradient with PyTorch Autograd***

```python
import torch

def grad_of_quadratic(x_value: float) -> float:
    # TODO: build a tracked leaf for x, compute f(x), run backprop, return df/dx as a float
    x = torch.tensor(x_value, dtype=torch.float32, requires_grad=True)
    f_x = x**2 + 3*x + 2
    f_x.backward()
    return x.grad.item()

print(grad_of_quadratic(2.0))  # 7.0
print(grad_of_quadratic(0.0))  # 3.0
print(grad_of_quadratic(-1.0)) # 1.0
```
